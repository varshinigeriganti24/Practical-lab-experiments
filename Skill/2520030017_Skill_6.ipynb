```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/placement_predict_50k_adjusted.csv")

# Fill missing numerical values with column mean
# inplace=True tells pandas to modify the original DataFrame directly,
# instead of creating a new DataFrame.
df.fillna(df.mean(numeric_only=True), inplace=True)

# Save the updated dataset to a new CSV file
# index=False tells pandas not to save the row numbers (index) into the CSV file.
df.to_csv("placementfilled.csv", index=False)

# Check if any missing values remain
print(df.isnull().sum())

print("\nNew dataset saved as 'plcementfilled.csv'")
```

```python
import pandas as pd

df = pd.read_csv("/content/placement_predict_50k_adjusted.csv")

# Fill missing numerical values with column mean
df.fillna(df.mean(numeric_only=True), inplace=True)

# Fill missing categorical values with mode
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

df.to_csv("placementfilledcat.csv", index=False)

print(df.isnull().sum())
```
