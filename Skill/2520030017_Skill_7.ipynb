```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("/content/placement_predict_50k Dataset (1).csv")

# Select only two numerical features
X = df[['CodingTestScore', 'AptitudeTestScore']]

# Apply Standard Scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(
    X_scaled,
    columns=['CodingTestScore', 'AptitudeTestScore']
)

print("Original Data")
print(X.head())

print("\nAfter Standard Scaling")
print(scaled_df.head())
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Load dataset
df = pd.read_csv("/content/placement_predict_50k Dataset (1).csv")

# Select only two numerical features
X = df[['CGPA', 'AptitudeTestScore']]

# Apply Min-Max Scaler
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(
    X_scaled,
    columns=['CGPA', 'AptitudeTestScore']
)

print("Original Data")
print(X.head())

print("\nAfter Min-Max Scaling")
print(scaled_df.head())
```
